```python
!pip install -U google-genai

from google import genai
from google.colab import userdata

# Initialize Gemini Client
client = genai.Client(api_key="")

MODEL = "gemini-3.5-flash"  # Standard model string for Gemini

faculty_data = {
    "Dr. Ayesha Malik": [
        "Software Engineering",
        "Software Project Management",
        "Software Quality Assurance"
    ],
    "Dr. Bilal Ahmed": [
        "Operating Systems",
        "Computer Networks",
        "Computer Architecture"
    ],
    "Dr. Hira Khan": [
        "Database Systems",
        "Database Management",
        "Data Warehousing"
    ],
    "Dr. Salman Raza": [
        "Artificial Intelligence",
        "Machine Learning",
        "Data Science"
    ],
    "Dr. Maryam Ali": [
        "Multivariable Calculus",
        "Linear Algebra",
        "Discrete Mathematics"
    ],
    "Dr. Hamza Farooq": [
        "Data Structures",
        "Programming Fundamentals",
        "Object Oriented Programming"
    ]
}

# Formatting faculty information string
faculty_information = ""
for faculty, courses in faculty_data.items():
    faculty_information += f"{faculty}: {', '.join(courses)}\n"

def fuuast_agent(question):
    prompt = f"""
You are a University Information AI Agent for Federal Urdu University of
Arts, Science and Technology (FUUAST).
Your purpose is to provide useful information about university teachers,
courses, and general university-related questions.

Faculty and course information available in the local database:
{faculty_information}

Instructions:
1. Use the local database when answering questions about teachers and
their assigned courses.
2. Do not create or modify teacher-course information.
3. If the requested teacher or course is not found in the database,
clearly mention that the information is not available in the current
database.
4. For general information about FUUAST, use your general knowledge when
appropriate.
5. Answer university-related questions clearly and briefly.
6. If the question is unrelated to the university, politely explain that
you are a university information agent.
7. Do not ask the user to provide information that you can answer using
the available knowledge.
8. Give direct and helpful answers.

User Question:
{question}

Answer:
"""
    result = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )
    return result.text

# Interactive CLI Loop
print("FUUAST University Information Agent is running.")
print("You can ask about faculty members, courses, or university information.")
print("Type 'exit' to close the agent.\n")

while True:
    question = input("User: ")
    if question.lower() == "exit":
        print("Agent closed. Thank you!")
        break
    try:
        response = fuuast_agent(question)
        print("\nAgent:", response, "\n")
    except Exception as error:
        print("Error:", error)
```
